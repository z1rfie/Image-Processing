```python
def gamma_correction(img, gamma):
    img_norm = img / 255.0
    corrected = np.power(img_norm, gamma)
    return (corrected * 255).astype(np.uint8)
```

```python
light = gamma_correction(image_gray, 0.3)
```

```python
plt.imshow(light, cmap='gray')
```

```python
dark = gamma_correction(image_gray, 3.0)
```

```python
plt.imshow(dark, cmap='gray')
```

```python
from skimage.metrics import structural_similarity, mean_squared_error

(ssim, diff) = structural_similarity(image_gray, dark, full=True)
diff = (diff * 255).astype("uint8")
print("SSIM: {}".format(ssim))
```

```python
plt.imshow(diff, cmap='gray')
```

```python
mse = mean_squared_error(image_gray, dark)
mse
```

```python
from skimage.metrics import structural_similarity, mean_squared_error

(ssim, diff) = structural_similarity(image_gray, light, full=True)
diff = (diff * 255).astype("uint8")
print("SSIM: {}".format(ssim))
```

```python
plt.imshow(diff, cmap='gray')
```

```python
mse = mean_squared_error(image_gray, light)
mse
```

```python
eq_gray = cv2.equalizeHist(image_gray)
```

```python
plt.imshow(eq_gray, cmap="gray")
```

```python
plt.imshow(image_gray, cmap="gray")
```

```python
Es = eq_gray.mean()
Ds = eq_gray.std()
Et = mean
Dt = std
```

```python
print(Es,Ds)
```

```python
stat_corrected = Es + (image_gray.astype(np.float64) - Et) * (Ds / Dt)
stat_corrected = np.clip(stat_corrected, 0, 255).astype(np.uint8)
```

```python
print(stat_corrected.mean(), stat_corrected.std())
```

```python
plt.imshow(stat_corrected, cmap='gray')
```
